```python
import numpy as np

text = ["I", "love", "deep", "learning"]
word_to_index = {"I": 0, "love": 1, "deep": 2, "learning": 3}
index_to_word = {i: w for w, i in word_to_index.items()}
sequence = [word_to_index[word] for word in text]
X = sequence[:3]
y = sequence[3]

vocab_size = len(word_to_index)
input_size = vocab_size
hidden_size = 8
output_size = vocab_size

np.random.seed(0)
Wxh = np.random.randn(hidden_size, input_size) * 0.01
Whh = np.random.randn(hidden_size, hidden_size) * 0.01
Why = np.random.randn(output_size, hidden_size) * 0.01
bh = np.zeros((hidden_size, 1))
by = np.zeros((output_size, 1))

def one_hot(idx, size):
    vec = np.zeros((size, 1))
    vec[idx] = 1
    return vec

learning_rate = 0.1
epochs = 500

for epoch in range(epochs):
    h_prev = np.zeros((hidden_size, 1))
    inputs = [one_hot(i, vocab_size) for i in X]

    for x in inputs:
        h_prev = np.tanh(np.dot(Wxh, x) + np.dot(Whh, h_prev) + bh)

    y_pred = np.dot(Why, h_prev) + by
    probs = np.exp(y_pred) / np.sum(np.exp(y_pred))
    loss = -np.log(probs[y][0])

    dy = probs
    dy[y] -= 1

    dWhy = np.dot(dy, h_prev.T)
    dby = dy
    dh = np.dot(Why.T, dy)

    for t in reversed(range(len(inputs))):
        dtanh = (1 - h_prev ** 2) * dh
        dWxh = np.dot(dtanh, inputs[t].T)
        dWhh = np.dot(dtanh, h_prev.T)
        dbh = dtanh

        Wxh -= learning_rate * dWxh
        Whh -= learning_rate * dWhh
        Why -= learning_rate * dWhy
        bh -= learning_rate * dbh
        by -= learning_rate * dby

    if epoch % 100 == 0:
        print(f"Epoch {epoch}, Loss: {loss:.4f}")

h = np.zeros((hidden_size, 1))
for x in [one_hot(i, vocab_size) for i in X]:
    h = np.tanh(np.dot(Wxh, x) + np.dot(Whh, h) + bh)

y_pred = np.dot(Why, h) + by
predicted_index = np.argmax(y_pred)

print("Predicted word:", index_to_word[predicted_index])
```
